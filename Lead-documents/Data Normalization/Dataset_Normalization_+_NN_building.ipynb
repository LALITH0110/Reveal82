```python
#Creating a neural

class Model(nn.Module):
  def __init__(self, in_features=11, h1=8, h2=9, out_features=4):
    super().__init__() #instantiate our nn.Module
    self.fc1 = nn.Linear(in_features, h1)#fully connected
    self.fc2 = nn.Linear(h1, h2)#moving through layers
    self.out = nn.Linear(h2, out_features)

  def forward(self, x):
    x = F.relu(self.fc1(x))#reclified linear unit(<0 make it 0, else use the value)
    x = F.relu(self.fc2(x))#reclified linear unit(<0 make it 0, else use the value)
    x = self.out(x)

    return x
```
